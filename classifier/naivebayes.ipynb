```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.pipeline import Pipeline
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import classification_report, confusion_matrix
```

```python
df = pd.read_csv("sentiment_clean.csv")
df.dropna(subset = ["text", "label"])
df = df[["text", "label"]]

df

```

```python
X = df["text"]
Y = df["label"]

X_train, X_test, Y_train, Y_test = train_test_split(X, 
                                                    Y,
                                                    test_size=0.2,      
                                                    random_state=1)
```

```python
pipeline = Pipeline([
    ("tfidf", TfidfVectorizer(
        stop_words="english",
        ngram_range=(1, 3), # include both single words, bigrams, trigrams (e.g. "net loss narrowed")
    )),
    ("nb", MultinomialNB()), # multinomial because we are working with discrete counts/frequency
])

model = pipeline.fit(X_train, Y_train)
```

```python
# Evaluate
y_pred = model.predict(X_test)
print(classification_report(Y_test, y_pred, target_names=["bearish (0)", "bullish (1)"]))
print(confusion_matrix(Y_test, y_pred))
```

```python
model.score(X_test, Y_test)
```
